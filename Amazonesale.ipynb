```python
df['Month']=df['Date'].dt.month_name()
```

```python
#Monthly sales Trend

monthly_sales=df.groupby('Month')['Amount'].sum()
monthly_sales.plot(kind='line')
plt.show()
```

```python
monthly_sales.plot( kind='line',figsize=(8,4),marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.grid(True)
plt.show()
```

```python
#PRODUCT ANALYSIS
df['Category']= df['Category'].astype('category')
df['Size']=df['Size'].astype('category')
df['Status']=df['Status'].astype('category')
df['Fullfilment']=df['Fulfilment'].astype('category')
```

```python
category_sales=df.groupby('Category')['Amount'].sum()
category_sales.sort_values(ascending=False)
```

```python
sns.barplot(x=category_sales.values,y=category_sales.index)
```

```python
#SIZE ANALYSIS
size_sales=df['Size'].value_counts()
size_sales.plot(kind='bar')
```

```python
#FULFILMENT ANALYSIS
fulfilment_sales=df.groupby('Fulfilment')['Amount'].sum()
print(fulfilment_sales)
```

```python
fulfilment_sales.plot(kind='pie',autopct='%1.1f%%')
```
